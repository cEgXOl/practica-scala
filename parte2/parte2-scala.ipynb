```scala
## Ejercicio 6 — Decidir el ganador de una partida

En este ejercicio se crea una función llamada ganador que compara dos
manos de cartas y determina cuál es la puntuación ganadora.

La función utilizará la función bust creada anteriormente para detectar
si alguna mano supera 21.

Reglas aplicadas:

1. Si ambas manos superan 21, no hay ganador y devuelve 0.
2. Si solamente la mano A supera 21, gana la mano B.
3. Si solamente la mano B supera 21, gana la mano A.
4. Si ninguna supera 21, gana la mano con mayor puntuación.
```

```scala
def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

def ganador(handA: Int, handB: Int): Int = {
    if (bust(handA) && bust(handB)){
        0
    } else if (bust(handA)){
        handB
    } else if (bust(handB)) {
        handA
    } else if (handA > handB) {
        handA
    } else {
        handB
    }
}

println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

```scala
### Conclusión
Entre 26 y 20 La mano A supera 21 se imprime 20 
Entre 18 y 22 La mano B supera 21 se imprime 18
Entre 24 y 25 Las manos superan 21 se imprime 0
Entre 17 y 19 Ninguna supera 21 se imprime 19
Entre 21 y 20 Ninguna supera 21 se imprime 21
```

```scala
## Ejercicio 9 — Recorrer un Array con while

En este ejercicio se recorre un Array utilizando un bucle while.

Para recorrer todas las posiciones del Array se utiliza una variable
contador que comienza en 0 y aumenta hasta alcanzar el tamaño del Array.

Este método representa una forma de recorrer Arrays.
```

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0

def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

while (i < manos.length) {
  println(s"La mano vale: ${manos(i)} y su resultado es: ${bust(manos(i))}")
  i = i + 1
}
```

```scala
### Conclusión

El contador es mutable para que cambie de valor cada vez que se ejecuta el bucle.
Si no se incrementara i sería un bucle infinito.
Corresponde al estilo imperativo el cambio de la variable i para controlar el flujo del bucle while.
```

```scala
## Ejercicio 10 — Listas e inmutabilidad

En este ejercicio se trabaja con listas (`List`) en Scala.

Las listas son estructuras inmutables, por lo que las operaciones que
parecen modificar una lista realmente generan una nueva lista sin alterar
la original.

Se comprobará este comportamiento utilizando el operador `::`.
```

```scala
val jugadores = List("Alex", "Chen", "Marta")

println(jugadores)

val nuevosJugadores = "Sindhu" :: jugadores

println(s"Lista antigua: $jugadores lista nueva: $nuevosJugadores")

println("Longitud de la lista original:")
println(jugadores.length)

println("Lista original invertida:")
println(jugadores.reverse)
```

```scala
### Conclusión

:: no modifica la lista original porque las listas en Scala son estructuras
inmutables, crea una nueva lista colocando el elemento indicado al
principio de la lista existente.
```

```scala
## Ejercicio 11 — Construcción y concatenación de listas

En este ejercicio se construyen listas utilizando Nil y el operador ::.

Posteriormente se concatenan dos listas mediante el operador :::.

Como las listas en Scala son estructuras inmutables, las operaciones crean
nuevas listas sin modificar las originales.
```

```scala
val lista1 = "Ana" :: "Luis" :: "Marta" :: Nil
val lista2 = "Pedro" :: "Sofia" :: Nil

println(s"Lista 1 $lista1")
println(s"Lista 2 $lista2")
println("Concatenando...")

val listasJuntas = lista1 ::: lista2

println(s"Lista 1 $lista1")
println(s"Lista 2 $lista2")
println(s"Lista concatenada $listasJuntas")
```

```scala
## Ejercicio 12 — Operadores relacionales y lógicos

En este ejercicio se utilizan operadores relacionales y lógicos para
evaluar diferentes condiciones sobre las puntuaciones de varias manos.

Los operadores relacionales comparan valores y devuelven un resultado
booleano (true o false).

Los operadores lógicos permiten combinar varias condiciones.
```

```scala
val handA = 18
val handB = 21
val handC = 25

def verdad(bool: Boolean): String = {
  if (bool) {
    "Si"
  } else {
    "No"
  }
}

println(s"¿Es handA mayor que handB? ${verdad(handA > handB)}")
println(s"¿Es handB igual a 21? ${verdad(handB == 21)}")
println(s"¿Es handC diferente de 21? ${verdad(handC != 21)}")
println(s"¿Son handA y handB menores o iguales a 21? ${verdad(handA <= 21 && handB <= 21)}")
println(s"¿Alguna de las manos handA o handC supera 21? ${verdad(handA > 21 || handC > 21)}")
println(s"¿No se ha pasado handB de 21? ${verdad(!(handB > 21))}")
```

```scala
### Conclusion

| Expresión | Operador utilizado | Resultado |
|---|---|---|
| handA > handB | > | false porque 18 no es mayor que 21 |
| handB == 21 | == | true porque 21 es igual a 21 |
| handC != 21 | != | true porque 25 es diferente de 21 |
| handA <= 21 && handB <= 21 | <= y && | true porque ambas manos cumplen la condición |
| handA > 21 || handC > 21 | > y || | true porque handC supera 21 |
| !(handB > 21) | ! | true porque handB no supera 21 |
```

```scala
## Ejercicio 13 — foreach y funciones como valores

En este ejercicio se comparan dos formas de recorrer un Array.

La primera versión utiliza un enfoque imperativo mediante un bucle while,
donde se controla manualmente el índice de recorrido.

La segunda versión utiliza foreach, pasando una función que se ejecuta
para cada elemento del Array.

La segunda aproximación se acerca más al estilo funcional de Scala.
```

```scala
val manos = Array(17, 24, 21, 26, 18)

def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

var i = 0
println("Prueba A")
while (i < manos.length) {

  val mano = manos(i)

  println(s"Mano: $mano se pasa: ${bust(mano)}")

  i = i + 1
}

println("Prueba B")

manos.foreach { mano =>

  println(s"Mano: $mano se pasa: ${bust(mano)}")

}
```

```scala
### Conclusión

While necesita contador y una variable para iterar.
La versión con foreach se aproxima más al estilo funcional presentado
en el material de Datacamp porque utiliza funciones como argumentos y evita
modificar el estado mediante variables mutables.
```

```scala
## Ejercicio 14 — Efectos secundarios y estilo de programación

En este ejercicio se comparan dos formas de crear funciones.

La primera función modifica una variable externa mediante un efecto
secundario.

La segunda función utiliza un enfoque más funcional: recibe valores como
parámetros y devuelve un nuevo resultado sin modificar el estado externo.
```

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(4)
println(total)
sumarAlTotal(4)
println(total)
sumarAlTotal(4)
println(total)

def sumar(a: Int, b: Int): Int = {
  a + b
}

val resultado1 = sumar(2, 5)
val resultado2 = sumar(8, 7)

println(s"Resultado 1: $resultado1")
println(s"Resultado 2: $resultado2")
println(s"Valor actual de total: $total")
```

```scala
###Conclusión 

Análisis de la primera función

La función sumarAlTotal modifica una variable situada fuera de su ámbito
local.

Esto ocurre porque total es una variable global para la función, es decir,
está definida fuera del bloque de la función pero puede ser utilizada dentro
de ella.

El efecto secundario producido es:

Se modifica el valor de total y la misma llamada a la función puede producir diferentes resultados
dependiendo del estado actual de la variable.

Comparación de funciones

| Característica | sumarAlTotal | sumar |
|---|---|---|
| Modifica variables externas | Sí | No |
| Devuelve un resultado calculado | Si | Sí |
| Utiliza efecto secundario | Sí | No |
| Estilo predominante | Imperativo | Funcional |
```

```scala
## Ejercicio 15 — Programa integrado: torneo de Twenty-One

En este ejercicio se desarrolla un pequeño torneo de Twenty-One aplicando
los conceptos trabajados anteriormente.

Se utilizan listas, arrays, funciones, estructuras condicionales y
recorridos de colecciones.

Los nombres de los jugadores se almacenan en una estructura inmutable List, mientras que las puntuaciones se almacenan en un Array[Int].
```

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

def verdad(bool: Boolean): String = {
  if (bool) {
    "Si"
  } else {
    "No"
  }
}

var i = 0
var mejor = 0
var mejorJugador = " "
while (i < jugadores.length) {
    val jugador = jugadores(i)
    val puntuacion = manos(i)
    
    if (!bust(manos(i)) && manos(i) > mejor) {
        mejor = manos(i)
        mejorJugador = jugadores(i)
        
    }
    
    println(s"$jugador tiene $puntuacion puntos ¿Se pasa? ${verdad(bust(puntuacion))}")

  i = i + 1
}

println(s"La mejor puntuación es la de $mejorJugador con $mejor")

println("Segunda ronda")

val manosRonda2 = Array(22, 19, 20, 21)
val ronda2 = jugadores.zip(manosRonda2)
var mejorRonda2 = 0
var mejorJugadorRonda2 = ""

ronda2.foreach { case (jugador, puntuacion) =>
    
    println(s"$jugador tiene $puntuacion puntos ¿Se pasa? ${verdad(bust(puntuacion))}")
    
    if (!bust(puntuacion) && puntuacion > mejorRonda2) {
    mejorRonda2 = puntuacion
    mejorJugadorRonda2 = jugador
    }

}

println(s"La mejor puntuación de la ronda 2 es de $mejorJugadorRonda2 con $mejorRonda2")
```

```scala
## Conclusión

Partes mutables
i, utilizada como contador del bucle.
mejor, utilizada para guardar la mejor puntuación encontrada.
mejorJugador, utilizada para guardar el nombre del jugador ganador.
Array[Int] de puntuaciones es mutable porque sus elementos pueden modificarse.

Partes inmutables
La lista de jugadores es inmutable, por lo que no se modifica durante la ejecución.
También las variables declaradas con val no pueden reasignarse.

Uso de funciones
bust(), que comprueba si una puntuación supera 21.
verdad(), que transforma un valor booleano en un texto.

Estructuras de control utilizadas
while para recorrer los jugadores en la primera ronda.
foreach para recorrer la segunda ronda.
if para comprobar condiciones y seleccionar la mejor puntuación.
Operadores lógicos como ! y &&.

Parte más cercana al estilo funcional
La segunda ronda se acerca más al estilo funcional porque utiliza foreach y evita usar un contador manual.

Parte más cercana al estilo imperativo
La primera ronda se acerca al estilo imperativo porque utiliza while y variables mutables (var) que cambian durante la ejecución.
```
